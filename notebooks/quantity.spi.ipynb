```spiral
:! hitpoints = ()
:! milliseconds = ()
:! tiles = ()
:! gold_coins = ()

:= hp (value : i32) : units::amount hitpoints = units::amount value

:= ms (value : i32) : units::amount milliseconds = units::amount value

:= tile_count (value : i32) : units::amount tiles = units::amount value

:= gp (value : i32) : units::amount gold_coins = units::amount value

:= healed_to_cap (health : i32) : i32 = units::value_of (units::capped (hp 150) (units::plus (hp health) (hp 60)))
```

#### same_unit_arithmetic (test)

```spiral
///- --test --symbol
///> c

:= ~health = 150i32
:= left = units::value_of (units::minus (hp health) (hp 40))
(left, healed_to_cap health, units::at_least (ms 2000) (ms (health * 10))) |> _assert_eq (110i32, 150i32, !+)
```
