#### damage (test)

```spiral
///- --test static --symbol

:< testing
```

```spiral
:= level_bonus (level : i32) : i32 =
    level / 5

:| fight_mode =
    | Offensive
    | Balanced
    | Defensive

:= attack_factor_percent (mode : fight_mode) : i32 =
    ?= mode ?|
    | Offensive => 100
    | Balanced => 75
    | Defensive => 50

:= max_melee_damage (mode : fight_mode) (attack_value : i32) (attack_skill : i32) (level : i32) : i32 =
    (85 * attack_factor_percent mode * attack_value * attack_skill * 2 + 100000) / 200000 + level_bonus level

:= max_melee_damage_offensive (attack_value : i32) (attack_skill : i32) (level : i32) : i32 =
    max_melee_damage Offensive attack_value attack_skill level
```

#### max_melee_matches_canary_double_formula (test)

```spiral
///- --test --symbol
///> c

:= ~value = 50i32
max_melee_damage_offensive value 80 100 |> _assert_eq 360i32
max_melee_damage_offensive (value - 8) 65 8 |> _assert_eq 233i32
```

#### max_melee_matches_canary_at_the_extremes (test)

```spiral
///- --test --symbol
///> c

:= ~value = 50i32
max_melee_damage_offensive (value - 49) 1 1 |> _assert_eq 0i32
max_melee_damage_offensive (value + 50) 120 500 |> _assert_eq 1120i32
```

#### max_melee_matches_canary_for_weak_attacks (test)

#### max_melee_matches_canary_for_high_levels (test)

```spiral
///- --test --symbol
///> c

:= ~value = 50i32
max_melee_damage_offensive (value - 2) 99 250 |> _assert_eq 454i32
max_melee_damage_offensive (value + 150) 130 1000 |> _assert_eq 2410i32
```

#### fight_mode_scales_max_melee_like_canary (test)

```spiral
///- --test --symbol
///> c

:= ~skill = 30i32
max_melee_damage Offensive 21 skill 8 |> _assert_eq 55i32
max_melee_damage Balanced 21 skill 8 |> _assert_eq 41i32
max_melee_damage Defensive 21 skill 8 |> _assert_eq 28i32
```

#### a_low_skill_balanced_hit_matches_canary (test)

```spiral
///- --test --symbol
///> c

:= ~skill = 30i32
max_melee_damage Balanced 21 (skill - 19) 8 |> _assert_eq 16i32
```

```spiral
///- --test --symbol
///> c

:= ~level = 9i32
level_bonus level |> _assert_eq 1i32
level_bonus (level + 1) |> _assert_eq 2i32
```
